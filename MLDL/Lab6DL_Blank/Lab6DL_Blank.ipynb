# Decision Tree Classifier

```python
import pandas as pd
```

```python
from sklearn.datasets import load_iris 
```

# Import iris data ser using sklearn

```python
df=load_iris()
```

```python
df
```

```python
df.feature_names
```

```python
df.target
```

# Importing the libraries

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier,plot_tree
```

# Divide the data into input and output

```python
X = df.data 
```

```python
y =df.target
```

```python
y
```

# Splitting the dataset into the Training set and Test set

```python
X_train, X_test, y_train, y_test=train_test_split(X,y,random_state=42)
```

```python
import numpy as np
```

# Fitting DecisionTreeClassifier   on dataset

```python
clf=DecisionTreeClassifier()
```

```python
clf.fit(X_train,y_train)
```

# Display Decision Tree

```python
plot_tree(clf)
```

# Predict the x_test 

```python
predict=clf.predict(X_test)
predict
```

```python
clf.score(X_test,y_test)
```

# Import diabetes.csv dataset 

```python
data=pd.read_csv('diabetes.csv')
```

```python
data
```

# Check the distribution of the target

```python
data.Outcome.value_counts()
```

# Replace zeros with NaN for specific columns

```python
cols=['Glucose','BloodPressure','SkinThickness','Insulin','BMI']
data[cols]=data[cols].replace(0,np.nan)
```

# Check for missing values

```python
data.isnull().sum()
```

# Fill missing values with median

```python
for col in cols:
    data[col].fillna(data[col].mean(), inplace=True)
```

```python
data.isnull().sum()
```

# Visualize Distributions

```python
import seaborn as sns

sns.pairplot(data.select_dtypes(include='number'))
```

# Convert Targer data into interger code

```python
data['Outcome'] = data['Outcome'].map({'Non Diabetic': 0, 'Diabetic': 1})
```

```python
data
```

# Divide the data into input and output

```python
x=data.iloc[::,:-1:]
```

```python
y=data['Outcome']
```

# Splitting the dataset into the Training set and Test set

```python
X_train, X_test, y_train, y_test=train_test_split(x,y,random_state=42)
```

# Create Model

```python
model=DecisionTreeClassifier()
```

# Fitting DecisionTreeClassifier   on dataset

```python
model.fit(X_train, y_train)
```

# Display Decision Tree

```python
plot_tree(model)
```

# Predict the x_test 

```python
predict=model.predict(X_test)
```

# Display Training Accuracy

```python
model.score(X_train,y_train)
```

# Display Test Accuracy

```python
model.score(X_test,y_test)
```

# Confusion Matrix

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
cm=confusion_matrix(y_test,predict)
```

```python
cm
```

```python
print(classification_report(y_test,predict))
```

```python
accuracy_score(y_test,predict)
```

# Save Model State

```python
import pickle
```

```python
with open('model.pkl','wb') as file:
    pickle.dump(model,file)
```

# Load Model from Disk

```python
with open('model.pkl','rb') as file:
    load_model = pickle.load(file)
```

```python
load_model.predict(X_test)
```

```python
cm=confusion_matrix(y_test,load_model.predict(X_test))
cm
```

```python
print(classification_report(y_test,load_model.predict(X_test)))
```

```python
print(accuracy_score(y_test,load_model.predict(X_test)))
```
