```python
def step_function(x):
    if x>=0:
        return 1
    else:
        return 0

print("step functione defined...")
```

# Part 1: the OR Gate

```python
#inputs
X_or=np.array([[0,0],[0,1],[1,0],[1,1]])

#target
y_or=np.array([0,1,1,1])

print("AND gate declared...")
```

# Step 2: Intialize Weight

```python
weight=np.array([0,0])
bais=0
lr=0.5

print(f"Intialize weight: {weight}")
print(f"Intialize Bais: {bais}")
```

# Step 3 : Manual Calculation

```python
input_sample=X_or[1]
target_value=y_or[1]

#1. Weight sum
z=np.dot(input_sample,weight)+bais
print(f"Math: ({input_sample[0]} * {weight[0]}) + ({input_sample[1]} * {weight[1]}) + {bais}")
print(f"Sum: (z): {z:.2f}")

#2. Activation
prediction=step_function(z)
print(f"prediction: {prediction}")

#3. Error
error=target_value-prediction
print(f"Error: {error}")
```

# Epoch 

```python
epochs=5


print("------Training---------")
for epoch in range(epochs):
    print(f"\nEPOCH-> {epoch+1}")

    for i in range(4):
        x=X_or[i]
        y=y_or[i]

        #prediction
        z=np.dot(x, weight)+bais
        pred=step_function(z)

        #error 
        error= y-pred

        #Learn
        if error != 0:
            weight=weight+(lr*error*x)
            bais=bais+(lr*error*1)
            print(f"-> MISTAKE on {x}.Expected{y},got{pred}.")
            print(f"->UPDATED Weights: {weight}, Bais: {bais:.2f}")
        else:
            print(f"Correct on {x}")

print(f"\nFinal Weight: {weight}")
print(f"Final Bais {bais}")
```
