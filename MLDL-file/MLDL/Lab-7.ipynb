# Do all pre processing on diabetes csv

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.model_selection import train_test_split
```

# Import diabetes.csv data ser using sklearn

```python
df = pd.read_csv('diabetes.csv')
```

```python
df
```

# Check the distribution of the target

```python
df['Outcome'].value_counts()
```

# Replace zeros with NaN for specific columns

```python
cols = ['Glucose', 'BloodPressure', 'SkinThickness','Insulin','BMI']
df[cols] = df[cols].replace(0, np.nan)
df
```

# Check for missing values

```python
df.isnull().sum()
```

# Fill missing values with median

```python
for col in cols:
    df[col].fillna(df[col].median(), inplace=True)
```

```python
df.isnull().sum()
```

# Convert Targer data into interger code

```python
df['Outcome'] = df['Outcome'].map({'Non Diabetic': 0, 'Diabetic': 1})
```

```python
df
```

# Visualize Distributions

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.pairplot(df.select_dtypes(include='number'),)
plt.show()
```

```python
df
```

# Divide the data into input and output

```python
x = df.iloc[::,:-1:]
```

```python
y = df['Outcome']
```

# import KNN Model

```python
from sklearn.neighbors import KNeighborsClassifier
```

# Convert all features into Z score

```python
from sklearn.preprocessing import StandardScaler
```

```python
scale = StandardScaler()
```

```python
x_sclaed = scale.fit_transform(x)
```

```python
x_sclaed
```

# Splitting the dataset into the Training set and Test set

```python
x_train, x_test, y_train, y_test = train_test_split(x_sclaed, y, test_size=0.3, random_state=42)
```

# Create KNN Model

```python
knn = KNeighborsClassifier(n_neighbors=5)
```

# Fitting KNN   on dataset

```python
knn.fit(x_train, y_train)
```

```python
knn.score(x_test, y_test)
```

# Display classification_report

```python
from sklearn.metrics import confusion_matrix,ConfusionMatrixDisplay,classification_report
```

```python
cm = confusion_matrix(y_test, knn.predict(x_test))
cm
```

```python
disp = ConfusionMatrixDisplay(cm)
disp.plot()
plt.show()
```

```python
print(classification_report(y_test, knn.predict(x_test)))
```

# Use Cross Validation to find best hyperparameter for KNN

```python
from sklearn.model_selection import cross_val_score
```

```python
from sklearn.model_selection import KFold
```

```python
k = KFold(n_splits=10)
```

```python
for train_index, test_index in k.split(x_sclaed):
    x_train_kf, x_test_kf = x_sclaed[train_index], x_sclaed[test_index]
    y_train_kf, y_test_kf = y[train_index], y[test_index]
    knn.fit(x_train_kf, y_train_kf)
    print(knn.score(x_test_kf, y_test_kf))
```

```python
scores = cross_val_score(knn, x_sclaed, y, cv=k)
scores
```
