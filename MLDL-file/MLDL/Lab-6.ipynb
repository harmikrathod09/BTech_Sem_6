# Decision Tree Classifier

```python
from sklearn.datasets import load_iris 
```

# Import iris data ser using sklearn

```python
data = load_iris()
```

```python
data
```

# Importing the libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn import  tree
```

# Divide the data into input and output

```python
x = data.data
y = data.target
```

```python
data.target
```

# Splitting the dataset into the Training set and Test set

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3,random_state=42)
```

# Fitting DecisionTreeClassifier   on dataset

```python
dtc = DecisionTreeClassifier()
```

```python
dtc.fit(x_train, y_train)
```

# Display Decision Tree

```python
tree.plot_tree(dtc)
plt.show()
```

# Predict the x_test 

```python
dtc.score(x_train, y_train)
```

```python
dtc.score(x_test, y_test)
```

# Import diabetes.csv dataset 

```python
df = pd.read_csv('diabetes.csv')
```

```python
df
```

# Check the distribution of the target

```python
df['Outcome'].value_counts()
```

# Replace zeros with NaN for specific columns

```python
cols = ['Glucose', 'BloodPressure', 'SkinThickness','Insulin','BMI']
df[cols] = df[cols].replace(0, np.nan)
df
```

# Check for missing values

```python
df.isnull().sum()
```

# Fill missing values with median

```python
for col in cols:
    df[col].fillna(df[col].mean(), inplace=True)
```

```python
df.isnull().sum()
```

# Visualize Distributions

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.pairplot(df.select_dtypes(include='number'))
plt.show()
```

# Convert Targer data into interger code

```python
df['Outcome'] = df['Outcome'].map({'Non Diabetic': 0, 'Diabetic': 1})
```

```python
df
```

# Divide the data into input and output

```python
x = df.iloc[::,:-1:]
```

```python
y = df['Outcome']
```

# Splitting the dataset into the Training set and Test set

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

# Create Model

```python
model = DecisionTreeClassifier()
```

# Fitting DecisionTreeClassifier   on dataset

```python
model.fit(x_train, y_train)
```

# Display Decision Tree

```python
tree.plot_tree(model)
plt.show()
```

# Predict the x_test 

```python
model.predict(x_test)
```

# Display Training Accuracy

```python
model.score(x_train, y_train)
```

# Display Test Accuracy

```python
model.score(x_test, y_test)
```

# Confusion Matrix

```python
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score
```

```python
cm = confusion_matrix(y_test, model.predict(x_test))
cm
```

```python
print(classification_report(y_test, model.predict(x_test)))
```

# Save Model State

```python
import pickle
```

```python
with open('diabetes_model.pkl', 'wb') as f:
    pickle.dump(model, f)
```

# Load Model from Disk

```python
with open('diabetes_model.pkl', 'rb') as f:
    loaded_model = pickle.load(f)
```

```python
loaded_model.predict(x_test)
```

```python
cm = confusion_matrix(y_test, loaded_model.predict(x_test))
cm
```

```python
print(classification_report(y_test, loaded_model.predict(x_test)))
```
